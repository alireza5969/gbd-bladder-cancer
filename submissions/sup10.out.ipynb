# GATHER checklist (Supporting File 10)

**Manuscript:** Estimated Burden of Bladder Cancer in the United States and Its States, 1990–2021: A Descriptive Analysis of the Global Burden of Disease Study 2021

**Reporting guideline:** Guidelines for Accurate and Transparent Health Estimates Reporting (GATHER) (Stevens et al., *Lancet* 2016;388:e19–e23).

**Nature of the study:** secondary, descriptive analysis of published GBD 2021 estimates. The authors did not synthesise primary data sources or fit the GBD models; items about data inputs and model building therefore refer to the GBD 2021 publications.
